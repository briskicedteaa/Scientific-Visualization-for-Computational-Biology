```python
import plotly.express as px
import pandas as pd
```

```python
df = pd.read_csv('/content/Ligands2.csv', sep =';' )
df = df.drop(columns=df.columns.difference(['Ligand Efficiency LE', 'Ligand Efficiency LLE', 'Ligand Efficiency BEI', 'Ligand Efficiency SEI', 'Molecular Weight', 'Standard Value', 'AlogP']))
df = df.dropna()
df
```

```python
from sklearn.decomposition import PCA

X = df[['Molecular Weight', 'AlogP', 'Standard Value', 'Ligand Efficiency BEI', 'Ligand Efficiency LE', 'Ligand Efficiency LLE', 'Ligand Efficiency SEI']]

pca = PCA(n_components=2)
X_reduced = components = pca.fit_transform(X)

PC1 = X_reduced[:, 0]
PC2 = X_reduced[:, 1]

total_var = pca.explained_variance_ratio_.sum() * 100

fig = px.scatter(x=PC1, y=PC2, color=X['Molecular Weight'],
                 title=f'Total Explained Variance: {total_var:.2f}%',
                 labels={'x': 'PC1', 'y': 'PC2'},
                 width=800, height=600,
                 render_mode='svg')

fig.update_traces(marker_line_color='black', marker_line_width=1)
fig.show()

fig.write_html("pca_scatter_plot.html")
```
