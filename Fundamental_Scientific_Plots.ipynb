```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv('/content/Ligands.csv', sep =';' )
df = df.drop(columns=df.columns.difference(['Ligand Efficiency LE', 'Ligand Efficiency LLE']))
df = df.dropna()
df
```

```python
sns.set_theme()
sns.displot(df, x="Ligand Efficiency LE")
```

```python
g = sns.relplot(data= df, x="Ligand Efficiency LLE", y="Ligand Efficiency LE")
g.ax.axline(xy1=(10, 2), slope=.2, color="b", dashes=(5, 2))
```
